```python
import numpy as np  
import pickle
import os
import sys

from metrics import make_pred_dict, do_eval, eged, get_f1_sym
```

```python
gt_path = f'../data/k400_mr345_val_min_change_duration0.3.pkl'
with open(gt_path, 'rb') as f:
    gt_dict = pickle.load(f)
vids = list(gt_dict.keys())
```

```python
pred_cnt = 5
pred_dict_list = []
f1s = 0
for i in range(pred_cnt):
    with open(f'../outputs/your_predicted_folder/model_pred_dict_ep-1.pkl', 'rb') as f:
        pred_dict = pickle.load(f)
        pred_dict = make_pred_dict(gt_dict, pred_dict, threshold=0.5)
        pred_dict_list.append(pred_dict)
        f1, _, _ = do_eval(gt_dict, pred_dict)
        f1s += f1
f1s /= pred_cnt

_, div = [np.round(x*100, 2) for x in eged(vids, gt_dict, pred_dict_list)]
f1_sym, f1_g2p, f1_p2g = [np.round(x*100, 2) for x in get_f1_sym(vids, gt_dict, pred_dict_list)]
f1s = np.round(f1s*100, 2)
print("F1_sym:", f1_sym)
print("F1_p2g:", f1_p2g)
print("F1_g2p:", f1_g2p)
print("Diversity:", div)
print(f"F1 over {pred_cnt}:", f1s)
```
